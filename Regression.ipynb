```python
import keras
from sklearn.preprocessing import StandardScaler
from keras.models import Sequential
from keras.layers import Dense
import numpy as np
```

## 2. Data collection

```python
(x_train,y_train),(x_test,y_test )= keras.datasets.boston_housing.load_data()
```

## 3. Data uderstanding

```python
x_train.shape,y_train.shape
```

```python
x_test.shape,y_test.shape
```

## 4. Data preparation

```python
std = StandardScaler()
```

```python
x_train = std.fit_transform(x_train)
x_test = std.fit_transform(x_test)
```

```python
x_train.std().round(),x_train.mean().round()
```

```python
x_test.std().round(),x_test.mean().round()
```

## 5. Model Building

```python
model = Sequential()
```

```python
model.add(Dense(100,activation="relu",input_shape=(13,)))
model.add(Dense(38,activation="relu"))
model.add(Dense(1,))
```

```python
model.compile(optimizer='adam',loss="mse",metrics=["mae"])
```

## 6. Model Training || 7. Model Testing || 8. Model Evaluation

```python
model.fit(x=x_train,y=y_train,batch_size=32,epochs=10,validation_split=0.2,verbose=1)
```

## 9. Model prediction

```python
new_data = np.array([[0.2,50,3.0,45,78,98,58,41,23,14,65,21,36]])
new_data = std.fit_transform(new_data)
prediction_out = model.predict(new_data)
print(prediction_out)
```

```python
model.save("House_pice.h5")
```
